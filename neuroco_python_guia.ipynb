## Sección 1: Variables y Tipos de Datos

Una **variable** es un contenedor donde guardamos información.

Los tipos de datos principales en Python son:
| Tipo | Nombre | Ejemplo |
|------|--------|---------|
| Número entero | `int` | `42` |
| Número decimal | `float` | `3.14` |
| Texto | `str` | `"corteza"` |
| Verdadero/Falso | `bool` | `True` |

````python
# Variables básicas

nombre_region = "Hipocampo"          # str (texto)
num_neuronas = 40_000_000            # int (entero)
potencial_reposo = -70.0             # float (decimal)
esta_activa = False                  # bool (booleano)

````

###¿Cómo mostramos la información?

La función `print()` sirve para mostrar información (texto, números, resultados de variables) en la consola o pantalla.

Probemos primero imprimir un Hola mundo!

````python
print("Hola Mundo!")  # Recuerda que las cadenas (texto) deben estar entre paréntesis
````

Ahora vamos a imprimir las variables que creamos anteriormente con `print()`

````python
# Imprimir las variables con print()
print("Región cerebral:", nombre_region)
print("Número de neuronas aproximado:", num_neuronas)
print("Potencial de reposo (mV):", potencial_reposo)
print("¿Está activa la neurona?", esta_activa)
````

También podemos ver el tipo de cualquier variable con `type()`

````python
# Podemos verificar el tipo de una variable con type()
print(type(nombre_region))    # <class 'str'>
print(type(num_neuronas))     # <class 'int'>
print(type(potencial_reposo)) # <class 'float'>
print(type(esta_activa))      # <class 'bool'>
````

## Sección 2: Operaciones Matemáticas

Python puede funcionar como una calculadora. Puede realizar operaciones básicas y avanzadas directamente usando operadores aritméticos.

| Operación | Símbolo |
|------|--------|
| Suma | `+` |
| Resta | `-` |
| Multiplicación | `*` |
| División | `/` |
| División entera | `//` |
| Módulo | `%` |
| Potencia | `**` |

````python
# Operaciones aritméticas
print("Suma:", 10 + 5)
print("Resta:", 10 - 5)
print("Multiplicación:", 10 * 5)
print("División:", 10 / 5)
print("División entera:", 10 // 3)
print("Módulo (residuo):", 10 % 3)
print("Potencia:", 2 ** 8)   # 2 elevado a la 8
````

Ahora probemos calcular algo sencillo:

Conversor de grados Celsius (°C) a grados Fahrenheit (°F):

💡Recuerda que la relación entre Celsius y Fahrenheit es:

$°F = (°C * \frac{9}{5}) + 32$

La función `input()` sirve para capturar datos introducidos por el usuario durante la ejecución de un programa

````python
celsius = input("Ingrese temperatura en grados Celcius:") # OJO: input guarda las variables como texto (str)

Fahrenheit = (int(celsius) * 9/5) + 32 # usar int() convierte las variables a números enteros
# print(f"") permite colocar variables entre el texto
# Pon una 'f' antes de las comillas y usa {} para insertar variables
print(f"{celsius} °C = {Fahrenheit} °F")
````

##  Sección 3: Listas

### Listas `[]`
Las listas almacenan múltiples valores en orden. Son perfectas para guardar series de tiempo, registros de señales, etc.



````python
# Creemos una lista de sujetos de un estudio clínico

nombres = ["Alice", "Bob", "Charlie", "David", "Eve"] # nombres de los sujetos del estudio
edades = [25, 30, 35, 40, 45] # edades de los sujetos del estudio

# Podemos ver el número de participantes con len()
print("Número de participantes:", len(nombres))

# Podemos ver el primer sujeto con índices
print("Primer sujeto:", nombres[0]) # índice 0 = primero
print("Último sujeto:", nombres[-1]) # índice -1 = último

# Podemos ver cuál es la mayor edad de los sujetos con max() y la mínima con
# min()
print("Edad máxima:", max(edades))
print("Edad mínima:", min(edades))
````

````python
# Operaciones con listas
regiones = ["Hipocampo", "Amígdala", "Biceps"]
print(regiones)

# Eliminar elemento con .remove()
regiones.remove("Biceps")
print("Sin biceps:", regiones)

# Agregar elemento con .append()
regiones.append("Tálamo")
print("Con Tálamo:", regiones)

# insert(índice, elemento): Inserta un elemento en una posición específica, desplazando los demás
regiones.insert(2, "Septo")
print("Con Septo:", regiones)

# extend(iterable): Añade todos los elementos de otra lista (o cualquier iterable) al final
regiones.extend(["Fórnix", "Cuerpo calloso" ])
print(regiones)

# Slicing: obtener una parte de la lista
print("Primeras 2 regiones:", regiones[0:2]) # toma desde la posición 0 hasta la posición 2

# Verificar si algo está en la lista con in
print("¿Está el Hipocampo? :", "Hipocampo" in regiones)
````

Los condicionales permiten que el programa tome decisiones según el valor de una variable.

## Sección 5: Ciclos (`for` y `while`)

Los ciclos repiten acciones.

`for` se usa para iterar sobre secuencias conocidas (listas, rangos, cadenas), mientras que `while` repite acciones basándose en una condición booleana hasta que esta sea falsa.

Ambos controlan el flujo para automatizar tareas repetitivas.

Usemos `for` para calcular el factorial de un número

image.png

````python
numero = int(input("Ingresa un número: "))
factorial = 1
for i in range(1, numero + 1): # range() comienza a contar desde 1 hasta el número ingresado más 1
  factorial = factorial * i

print("El factorial de", numero, "es", factorial)
````

Usemos `while` para crear un contador

````python
contador = 1 # donde va a comenzar a contar
limite= int(input("¿Hasta dónde quieres contar?: "))
while contador <= limite: # la condición del ciclo es que sea menor igual al limite
    print("Número:", contador)
    contador = contador + 1 # va a contar de uno en uno
````

## Sección 6: Funciones

Las funciones son bloques de código reutilizables. En lugar de repetir código, defines una función una vez y la llamas cuando la necesitas.

```python
def nombre_funcion(parametros):
    # código
    return resultado
```

Creemos una función para calcular el área de un rectángulo

````python
# Definir la función

def calcular_area(base, altura):
    area = base * altura
    return area
````

````python
# Llamar la función
b = float(input("Ingresa la base del rectángulo (cm): "))
h = float(input("Ingresa la altura del rectángulo (cm): "))

area_rectangulo = calcular_area(b, h)
print(f"El área del rectángulo es: {area_rectangulo} cm")
````

## Sección 7: Librerías

**NumPy** es la librería fundamental para cálculo numérico en Python. **Matplotlib** es la librería principal para graficar datos.

Son herramientas esenciales para trabajar con datos como señales.

````python
# Importar librerías (SIEMPRE ejecuta esta celda primero)
# as sirve para asignar un alias o nombre alternativo a un módulo, biblioteca o función al importarlo.
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Datos
x = np.array([1, 2, 3, 4, 5]) # vector de números del 1 al 5
y = x * 2  # relación simple

# Crear gráfica
plt.plot(x, y) # primero va el eje x y luego el eje y

# Etiquetas
plt.title("Función lineal") # title le asigna el título a la gráfica
plt.xlabel("Eje X") # xlabel le asigna el título al eje x
plt.ylabel("Eje Y") # ylabel le asigna el título al eje y
plt.grid(True) # crea una cuadrícula en la gráfica

# Mostrar
plt.show()
````

````python
# Arrays de NumPy: como listas pero mucho más potentes para cálculos
voltajes = np.array([-70.0, -68.5, -62.0, -54.0, 30.0, -80.0, -70.0])

print("Array:", voltajes)
print("Media:", np.mean(voltajes))
print("Desviación estándar:", np.std(voltajes))

# Operaciones vectorizadas (mucho más rápido que un ciclo for)
voltajes_V = voltajes / 1000  # convertir todo el array a Volts de una vez
print("En Volts:", voltajes_V)
````

````python
# Graficar una función senoidal

# Datos
tiempo = np.linspace(0, 10, 100) # 100 puntos entre 0 y 10
seno = np.sin(tiempo) # np.sin permite crear una funcion senoidal unitaria

# Graficar
plt.plot(tiempo, seno, color="red", linestyle="--", marker="o", label="Seno")
plt.title("Función senoidal")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")

# Leyenda
plt.legend()

# Cuadrícula
plt.grid(True)

# Mostrar
plt.show()
````

Vamos a divertirnos un poco, visualicemos una imagen de resonancia magnética :D

````python
import nibabel as nib

# Cargar la imagen
ruta = "/content/sub-002_ses-1.5T_task-3dffe_run-1_T1w.nii"
imagen = nib.load(ruta)

# Obtener los datos como array
data = imagen.get_fdata()

print("Dimensiones de la imagen:", data.shape)

# Cortes centrales (corte que se encuentra en la mitad de los datos)
x = data.shape[0] // 2  # sagital
y = data.shape[1] // 2  # coronal
z = data.shape[2] // 2  # axial

# Crear figura con 3 subplots
plt.figure(figsize=(12, 4))

# Plano sagital
plt.subplot(1, 3, 1)
plt.imshow(data[x, :, :].T, cmap='gray', origin='lower')
plt.title("Sagital")
plt.axis('off')

# Plano coronal
plt.subplot(1, 3, 2)
plt.imshow(data[:, y, :].T, cmap='gray', origin='lower')
plt.title("Coronal")
plt.axis('off')

# Plano axial
plt.subplot(1, 3, 3)
plt.imshow(data[:, :, z].T, cmap='gray', origin='lower')
plt.title("Axial")
plt.axis('off')

plt.tight_layout()
plt.show()
````

````python
from ipywidgets import interact

def ver_slice(slice_idx):
    # Crear figura con 3 subplots
    plt.figure(figsize=(12, 4))

    # Plano sagital
    plt.subplot(1, 3, 1)
    plt.imshow(data[slice_idx, :, :].T, cmap='gray', origin='lower')
    plt.title("Sagital")
    plt.axis('off')

    # Plano coronal
    plt.subplot(1, 3, 2)
    plt.imshow(data[:, slice_idx, :].T, cmap='gray', origin='lower')
    plt.title("Coronal")
    plt.axis('off')

    # Plano axial
    plt.subplot(1, 3, 3)
    plt.imshow(data[:, :, slice_idx].T, cmap='gray', origin='lower')
    plt.title("Axial")
    plt.axis('off')


interact(ver_slice, slice_idx=(0, data.shape[2]-1))
````
